```python
# plot mean and standard deviation: 1 sigma
def plot_mean_std(ax, cdfs, l, c, text_size=12, axhline_y=None):
    sigma =1
    mean = np.nanmean(cdfs, axis=0)
    std = np.nanstd(cdfs, axis=0)
    ax.plot(mean, label=l, color=c)
    ax.fill_between(np.arange(0, 100), mean - sigma*std, mean + sigma*std, alpha=0.2)
    ax.tick_params(axis='both', labelsize=text_size)
    if axhline_y is not None:
        ax.axhline(y=axhline_y, c='darkgray', xmin=0, xmax=100, linewidth=2, linestyle='--', zorder=-1)
```

```python
def plot_mean_std_dist(ax, dist, l, c, text_size=12):
    sigma = 1
    mean = np.nanmean(dist, axis=(0,2))
    std = np.nanstd(dist, axis=(0,2))
    ax.plot(np.arange(1, 101), mean, label=l, color=c)
    ax.fill_between(np.arange(1, 101), mean - sigma*std, mean + sigma*std, alpha=0.2)
    ax.tick_params(axis='both', labelsize=text_size)

    
text_size=12
years=np.arange(2010,2022)
spring_dist = np.zeros((p_total, t_days, len(years)*3))
summer_dist = np.zeros(spring_dist.shape)
autumn_dist = np.zeros(spring_dist.shape)
winter_dist = np.zeros(spring_dist.shape)

fig, axs = plt.subplots(nrows=1, ncols=4, sharex=True, sharey=True, figsize=(12,6))

mode1='2D'
mode2='3D'
for index, year in enumerate(years):
    dist_data = np.load(output_folder+"Distances/{0}-{1}/all_sep_array_{0}_{1}_{2}_tc{3}km.npy".format(mode1, mode2, year, threshold_dist))
    spring_dist[:, :, index*3:index*3+3] = dist_data[:, :t_days, spring_ind]
    summer_dist[:, :, index*3:index*3+3] = dist_data[:, :t_days, summer_ind]
    autumn_dist[:, :, index*3:index*3+3] = dist_data[:, :t_days, autumn_ind]
    winter_dist[:, :, index*3:index*3+3] = dist_data[:, :t_days, winter_ind]
    
plot_mean_std_dist(axs[0], spring_dist, str(mode1+'-'+mode2), new_cmap(0), text_size)
plot_mean_std_dist(axs[1], summer_dist, str(mode1+'-'+mode2), new_cmap(0), text_size)
plot_mean_std_dist(axs[2], autumn_dist, str(mode1+'-'+mode2), new_cmap(0), text_size)
plot_mean_std_dist(axs[3], winter_dist, str(mode1+'-'+mode2), new_cmap(0), text_size) 

axs[0].set_ylabel("Distance (km)" , fontsize=text_size)
axs[3].legend(loc="upper left")

axs[0].set_title("Austral spring \n (SON)", fontsize=text_size)
axs[1].set_title("Austral summer \n (DJF)", fontsize=text_size)
axs[2].set_title("Austral autumn \n (MAM)", fontsize=text_size)
axs[3].set_title("Austral winter \n (JJA)", fontsize=text_size)
axs[0].set_xlabel("Days", fontsize=text_size)
axs[1].set_xlabel("Days", fontsize=text_size)
axs[2].set_xlabel("Days", fontsize=text_size)
axs[3].set_xlabel("Days", fontsize=text_size)

axs[0].set_ylim(-10,300)
plt.subplots_adjust(hspace=0.1, wspace=0.1)

plt.savefig(output_folder+"Figures/Seasonal_separation_distance_2D-3D.pdf")
```

```python
color1 = cm.RdBu_r
def mapplot_diff(ax, array, threshold_dist): 
    gl = ax.gridlines(draw_labels=True)
    gl.xlines = False
    gl.xlocator = mticker.FixedLocator(x_ticks)
    gl.ylines = False
    gl.ylocator = mticker.FixedLocator(y_ticks)
    gl.top_labels = False
    gl.right_labels = False
    gl.bottom_labels = False
    gl.left_labels = False
    gl.xlabel_style = {'size': ticks_size, 'color': 'k'}
    gl.ylabel_style = {'size': ticks_size, 'color': 'k'}
    colormap = colors.ListedColormap(['gainsboro', 'white'])
    ax.set_xticks(x_ticks)
    ax.set_yticks(y_ticks)

    ax.pcolormesh(mask_lon[0, 1249:1750, 1499:], mask_lat[0, 1249:1750, 1499:], mask_land[0,0, 1250:1750, 1500:], cmap=colormap)
    
    ax.set_xlim(min_lon, max_lon)
    ax.set_ylim(min_lat, max_lat)
    ax.tick_params(axis='both',labelbottom=False, labeltop=False, labelleft=False, labelright=False)
    im = ax.scatter(coords['Longitude'], coords['Latitude'], c= array, s=scatter_size, cmap=color1, vmin=-60, vmax=60, marker='o',edgecolors='none')    
    return im


dvm_base=dvm_types[0]
i=1
dvm_mode = dvm_types[i]
fig,axes = plt.subplots(figsize=(8,5), dpi=dpi, nrows=1, ncols=4, sharex=True, sharey=True, subplot_kw={'projection': ccrs.PlateCarree()})

file=output_folder + "Distances/{0}-{1}/all_sep_array_{0}_{1}_{2}_tc{3}km.npy".format(dvm_base, other_mode, year, threshold_dist)
print(file)
dist_data_1 = np.load(file)

file2=output_folder+"Distances/{0}-{1}/all_sep_array_{0}_{1}_{2}_tc{3}km.npy".format(dvm_mode, other_mode, year, threshold_dist)
print(file2)
dist_data_2 = np.load(file2)

axes[0].set_ylabel(str(dvm_type_labels[i]+'-vs-'+other_mode +' $-$ '+ dvm_type_labels[0]+'-vs-'+other_mode), rotation=90, labelpad=5, fontsize=custom_size)

for j, mon in enumerate(select_months):
    # dist_data_1 = DVMI-2D, dist_data2 = DVMII-2D
    #diff=dist_data_2 - dist_data_1
    diff = utils.threshold_days(dist_data_2[:,:,mon],threshold_dist) - utils.threshold_days(dist_data_1[:,:,mon],threshold_dist)
    diff_mean = np.nanmean(diff)
    print('mean-diff'+str(dvm_type_labels[i]+'-vs-'+other_mode +' $-$ '+ dvm_type_labels[0]+'-vs-'+other_mode), diff_mean)

    mon_name = date(1900, mon+1, 1).strftime('%B')
    print(mon, mon_name, year)
    im = mapplot_diff(axes[j], diff, diff_mean)
    axes[j].set_title(mon_name+' '+str(year), fontsize=custom_size)
    axes[j].xaxis.set_tick_params(labelbottom=True, labelsize=custom_size)
    axes[j].xaxis.set_major_formatter(cticker.LongitudeFormatter())
    axes[j].text(19.5, -18.5, r'$\overline{diff} = $'+ str(np.round(diff_mean,1)) + ' d', fontsize=8, horizontalalignment='right')


axes[0].yaxis.set_tick_params(labelleft=True)
axes[0].yaxis.set_major_formatter(cticker.LatitudeFormatter())
axes[0].tick_params(axis='both', labelleft=True, labelsize=custom_size)
scale_bar(axes[j], 100, location=(0.78, 0.92),font_size=10)


cbar = fig.colorbar(im, ax=axes, orientation='horizontal', extend='both', shrink=0.6, pad=0.07)
cbar.set_label('Days', size= custom_size)
cbar.ax.tick_params(labelsize=custom_size)

plt.savefig(output_folder+"Figures/DVMII_vs_2D_diff_ThresholdDays_100km_2017.jpeg",  bbox_inches ='tight')
```
